# Null Handling

```python
import pandas as pd 
#import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\acer\Downloads\agriculture_data_with_nulls_duplicates.csv")
```

```python
print(df)
```

```python
# 1) Dropna    ---> remove entire rows having atleast one null
```

```python
print(df.dropna()) 
```

```python
# 2) dropna subset     ---> removes rows where name is null
```

```python
print(df.dropna(subset=['Farming_Type']))
```

```python
# 3) dropna how ='all'   ---> removes rows only when all columns are null
```

```python
print(df.dropna(how ='all'))
```

```python
# 4) dropna axis = 1     --->  removes columns containing nulls
```

```python
print(df.dropna(axis=1)) 
```

```python
# 5)  fillna constant          ---> replaces nulls with constant value
```

```python
print(df['Farming_Type'].fillna('unknown'))
```

```python
# 6) fillna mean               ---> numerial imputation
```

```python
print(df['Area_Acres'].fillna(df['Area_Acres'].mean()))
```

```python
#  7) fillna median                 ---> robust against outliers
```

```python
print(df['Yield_Ton_per_Acre'].fillna(df['Yield_Ton_per_Acre'].median()))
```

```python
# 8)  fillna mode                    ---> categorical imputation
```

```python
print(df['Income_INR'].fillna(df['Income_INR'].mode()[0]))
```

```python
# 9)  forward fill          ---> uses previous row value (df.fillna(method='ffill') or df.ffill()
```

```python
# print(df.fillna(method='ffill'))
print(df.ffill())
```

```python
# 10) backward fill         ---> uses next row value 
```

```python
# print(df.fillna(method='bfill')) 
print(df.bfill())
```

```python
# 11) replace()              ---> alternative null replacement 
```

```python
print(df.replace(np.nan,'unknown'))
```

```python
# 12) detect duplicates     ---> detect duplicate row mask
```

```python
print(df.duplicated())
```

```python
# 13) count duplicates      ---> number of duplicate rows
```

```python
print(df.duplicated().sum())
```

```python
# 14) remove duplicates     ---> removes duplicate rows
```

```python
print(df.drop_duplicates())
```
