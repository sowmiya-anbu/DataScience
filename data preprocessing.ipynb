### outlier detection

```python
plt.figure(figsize =(16,4))
sns.boxplot(x = df["Age"],color="green")
plt.title("Box Plot Of Age Outliers")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize =(12,2))
sns.boxplot(x = df["MonthlyIncome"],color="yellow")
plt.title("Box Plot Of MonthlyIncome Outliers")
plt.xlabel("MonthlyIncome")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize =(14,2))
sns.boxplot(x = df["CreditScore"],color="purple")
plt.title("Box Plot Of CreditScore Outliers")
plt.xlabel("CreditScore")
plt.ylabel("Frequency")
plt.show()
```

```python
#plt.figure(figsize =(10,3))
sns.boxplot(x = df["YearsExperience"],color="pink")
plt.title("Box Plot Of YearsExperience Outliers")
plt.xlabel("YearsExperience")
plt.ylabel("Frequency")
plt.show()
```

```python
# Age
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)
IQR = Q3 - Q1

age_outliers = df[(df["Age"] < (Q1 - 1.5*IQR)) | (df["Age"] > (Q3 + 1.5*IQR))]
print("Age Outliers")
print(age_outliers)
print("Count:", len(age_outliers))
```

```python
# MonthlyIncome
Q1 = df["MonthlyIncome"].quantile(0.25)
Q3 = df["MonthlyIncome"].quantile(0.75)
IQR = Q3 - Q1

income_outliers = df[(df["MonthlyIncome"] < (Q1 - 1.5*IQR)) | (df["MonthlyIncome"] > (Q3 + 1.5*IQR))]
print("MonthlyIncome Outliers")
print(income_outliers)
print("Count:", len(income_outliers))
```

```python
# CreditScore
Q1 = df["CreditScore"].quantile(0.25)
Q3 = df["CreditScore"].quantile(0.75)
IQR = Q3 - Q1

credit_outliers = df[(df["CreditScore"] < (Q1 - 1.5*IQR)) | (df["CreditScore"] > (Q3 + 1.5*IQR))]
print("CreditScore Outliers")
print(credit_outliers)
print("Count:", len(credit_outliers))
```

```python
# YearsExperience
Q1 = df["YearsExperience"].quantile(0.25)
Q3 = df["YearsExperience"].quantile(0.75)
IQR = Q3 - Q1

experience_outliers = df[(df["YearsExperience"] < (Q1 - 1.5*IQR)) | (df["YearsExperience"] > (Q3 + 1.5*IQR))]
print("YearsExperience Outliers")
print(experience_outliers)
print("Count:", len(experience_outliers))
```

```python
bins = [18,30,40,50,60]

labels = ["Young","Adult","Middle Age","Senior"]

df["Age_Group"] = pd.cut(
    df["Age"],
    bins = bins,
    labels = labels
    )
df["Age_Group"]
```

### Encoding

#### 1) Label encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["Gender"] = le.fit_transform(df["Gender"])

df.head()
```

#### 2) One-Hot Encoding

```python
encoded = pd.get_dummies(df["Gender"])

encoded
```

#### 3) Ordinal Encoding

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder()

df["Education"] = oe.fit_transform(df[["Education"]])

df.head()
```

#### 4) Binary Encoding

```python
from category_encoders import BinaryEncoder

be = BinaryEncoder(cols=["City"])

df = be.fit_transform(df)

df.head()
```

#### 5) Frequency Encoding 

```python
freq = df["City"].value_counts()

df["City"] = df["City"].map(freq)

df.head()
```

#### 6) Target Encoding

```python
from category_encoders import TargetEncoder

te = TargetEncoder(cols=["City"])

df["City"] = te.fit_transform(
    df["City"],
    df["Purchased"]
)
```
