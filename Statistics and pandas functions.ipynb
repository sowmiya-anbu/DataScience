```python
# -----> TO FIND STANDARD DEVIATION AND VARIANCE USING PYTHON <-----
```

```python
import pandas as pd
```

```python
x = [12,28,35,49,56,63,82,97,74]
```

```python
mean = sum(x) / len(x)
```

```python
numerator = 0 
for i in x:
    numerator += (i-mean) ** 2
```

```python
denominator = len(x)
```

```python
variance = numerator/denominator
```

```python
standard_deviation = variance ** 0.5
```

```python
print("MEAN =", mean) 
print("NUMERATOR =", numerator) 
print("DENOMINATOR =", denominator) 
print("VARIANCE =", variance)
print("STANDARD DEVIATION =", standard_deviation)
```

```python
# 8) OUTLIER DETECTION(IQR METHOD)
```

```python
x = np.array([5,10,15,20,25,30,35,40,45,50,55,70,65,60,75,80,85,90,95,100])
```

```python
Q1 = np.percentile(x,25)
Q3 = np.percentile(x,75)

IQR = Q3-Q1 

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR 

outliers = x[(x < lower) | (x > upper)]
```

```python
print("Q1:" , Q1)
print("Q3:" , Q3) 
print("IQR:" , IQR)
print("LOWER:" , lower )
print("UPPER:" , upper)
print("OUTLIERS:" , outliers)
```

```python
import pandas as pd
import numpy as np
```

```python
# Sample data
data = [12, 15, 18, 20, 22, 25, 30, 35, 40, 100]
```

```python
df = pd.DataFrame(data, columns=["Marks"])
```

```python
print(df)
```

```python
# Mean
print("Mean:", df["Marks"].mean())
```

```python
# Median
print("Median:", df["Marks"].median())
```

```python
# Mode
print("Mode:")
print(df["Marks"].mode())
```

```python
# Minimum
print("Minimum:", df["Marks"].min())
```

```python
# Maximum
print("Maximum:", df["Marks"].max())
```

```python
# Sum
print("Sum:", df["Marks"].sum())
```

```python
# Count
print("Count:", df["Marks"].count())
```

```python
# Range
print("Range:", df["Marks"].max() - df["Marks"].min())
```

```python
# Variance
print("Variance:", df["Marks"].var())
```

```python
# Standard Deviation
print("Standard Deviation:", df["Marks"].std())
```

```python
# Percentiles
print("25th Percentile:", df["Marks"].quantile(0.25))
print("50th Percentile:", df["Marks"].quantile(0.50))
print("75th Percentile:", df["Marks"].quantile(0.75))
```

```python
# Quartiles
Q1 = df["Marks"].quantile(0.25)
Q2 = df["Marks"].quantile(0.50)
Q3 = df["Marks"].quantile(0.75)

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
```

```python
# Interquartile Range (IQR) 
Q1 = df["Marks"].quantile(0.25)
Q2 = df["Marks"].quantile(0.50)
Q3 = df["Marks"].quantile(0.75)

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)

IQR = Q3 - Q1
print("IQR:", IQR)
```

```python
# Outlier Detection
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

outliers = df[(df["Marks"] < lower_limit) | (df["Marks"] > upper_limit)]
print("Outliers:",outliers)
```
