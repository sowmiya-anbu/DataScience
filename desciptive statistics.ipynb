### <------- Kurtosis ------->  3 Types

                Formula:


                                   ∑ ( x − μ )^4 / n 
                     Kurtosis = -------------------------         
                                 ( ∑ ( x − μ )^2 / n )^2

                Where:

                     x = Data values

                     μ = Mean

                     n = Number of observations

# 1) Leptokurtic Distribution (Kurtosis > 3)

     1) Definition:

            
             A Leptokurtic distribution has:

      
                 A high, sharp peak.

                 Heavy tails (more extreme values/outliers).

                 Kurtosis > 3.
            
     2) Formula:


                            ∑( x − μ )^4 / n ) 
             Kurtosis =  --------------------------
                          ( ∑( x − μ )^2 / n )^2

     Where:

        x = Data values
      
        μ = Mean
      
        n = Number of observations
      
        (x−μ)^2 = Second Moment (Variance)
     
        (x−μ)^4 = Fourth Moment

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Generate Leptokurtic Data
np.random.seed(10)
data = np.random.laplace(50,5,1000)

# Mean
mean = np.mean(data)

# Sample Size
n = len(data)

# Second Moment
m2 = sum(( x - mean ) ** 2 for x in data) / n

# Fourth Moment
m4 = sum(( x - mean ) ** 4 for x in data) / n

# Kurtosis
kurtosis = m4 / ( m2 ** 2)

print("Mean :", mean)
print("Sample Size :", n)
print("Second Moment :", m2)
print("Fourth Moment :", m4)
print("Kurtosis :", kurtosis)

if kurtosis > 3:
    print("Leptokurtic Distribution") 

plt.figure(figsize=(8,5))

# Histogram (Bar Plot)
sns.histplot(data,
             bins=30,
             stat="density",
             color="skyblue",
             edgecolor="black")

# KDE Curve
sns.kdeplot(data,
            color="red",
            linewidth=3)

plt.title("Leptokurtic Distribution")
plt.xlabel("Values")
plt.ylabel("Density")

plt.grid(True)
plt.show()
```

Explanation :

       The generated data follows a Laplace distribution, which naturally has heavier tails than a normal distribution.
           
       First, we calculate the mean.
           
       Next, we compute the second moment (variance) and the fourth moment.
           
       Using the kurtosis formula, we obtain the kurtosis value.
           
       Since the calculated kurtosis is greater than 3, the distribution is classified as Leptokurtic Distribution.
           
       Bar Plot: Bars are concentrated near the center with a high peak.
       
       Curve: Tall and narrow peak with heavy tails.

# 2) Mesokurtic Distribution (Kurtosis ≈ 3) 

         1) Definition:

                A Mesokurtic distribution has:

                     A moderate peak.
                         
                     Normal tails.
                         
                     Kurtosis = 3 (approximately).

         2) Formula:


                              ∑( x − μ )^4 / n ) 
                Kurtosis = --------------------------
                             ( ∑( x − μ )^2 / n )^2
     

         Where:

             x = Data values
  
             μ = Mean
  
             n = Number of observations
  
             (x−μ)^2 = Second Moment (Variance)
 
             (x−μ)^4 = Fourth Moment

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# Generate Mesokurtic Data
np.random.seed(10)
data = np.random.normal(50,10,1000)

# Mean
mean = np.mean(data)

# Sample Size
n = len(data)

# Second Moment
m2 = sum(( x - mean ) ** 2 for x in data) / n

# Fourth Moment
m4 = sum(( x - mean ) ** 4 for x in data) / n

# Kurtosis
kurtosis = m4 / (m2 ** 2)

print("Mean :", mean)
print("Sample Size :", n)
print("Second Moment :", m2)
print("Fourth Moment :", m4)
print("Kurtosis :", kurtosis)

if 2.8 <= kurtosis <= 3.2:
    print("Mesokurtic Distribution")

plt.figure(figsize=(8,5))

# Histogram (Bar Plot)
sns.histplot(data,
             bins=30,
             stat="density",
             color="lightgreen",
             edgecolor="black")

# KDE Curve
sns.kdeplot(data,
            color="blue",
            linewidth=3)

plt.title("Mesokurtic Distribution")
plt.xlabel("Values")
plt.ylabel("Density")

plt.grid(True)
plt.show()
```

Explanation :

        The data is generated using a Normal Distribution.
            
        The kurtosis is calculated using the mathematical formula.
            
        Since the kurtosis value is approximately 3, it is classified as a Mesokurtic Distribution.

        Bar Plot: Bars are moderately distributed around the center.
            
        Curve: Moderate bell-shaped peak.

# 3) Platykurtic Distribution (Kurtosis < 3)

      1) Definition :

               A Platykurtic distribution has:

                   A flat and broad peak.
                
                   Light tails.
                   
                   Kurtosis < 3.

      2) Formula :

            
                              ∑( x − μ )^4 / n ) 
               Kurtosis = --------------------------
                             ( ∑( x − μ )^2 / n )^2
     



       Where:

            x = Data values

            μ = Mean

            n = Number of observations

            (x−μ)^2 = Second Moment (Variance)
 
            (x−μ)^4 = Fourth Moment

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Generate Platykurtic Data
np.random.seed(10)
data = np.random.uniform(20,80,1000)

# Mean
mean = np.mean(data)

# Sample Size
n = len(data)

# Second Moment
m2 = sum(( x - mean ) ** 2 for x in data) / n

# Fourth Moment
m4 = sum(( x - mean ) ** 4 for x in data) / n

# Kurtosis
kurtosis = m4 / ( m2 ** 2)

print("Mean :", mean)
print("Sample Size :", n)
print("Second Moment :", m2)
print("Fourth Moment :", m4)
print("Kurtosis :", kurtosis)

if kurtosis < 3:
    print("Platykurtic Distribution")

plt.figure(figsize=(8,5))

# Histogram (Bar Plot)
sns.histplot(data,
             bins=30,
             stat="density",
             color="orange",
             edgecolor="black")

# KDE Curve
sns.kdeplot(data,
            color="green",
            linewidth=3)

plt.title("Platykurtic Distribution")
plt.xlabel("Values")
plt.ylabel("Density")

plt.grid(True)
plt.show()
```

Explanation : 

       The data is generated using a Uniform Distribution.
           
       The kurtosis is calculated using the mathematical formula.
           
       Since the kurtosis value is less than 3, it is classified as a Platykurtic Distribution.

       Bar Plot: Bars are spread widely with a low, flat peak.
           
       Curve: Flat and wide peak with light tails.
