# 1) Data Loading (Available Data) ----> Null Check

```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\acer\Downloads\diabetes - Copy.csv")
```

```python
print(df)
```

# Null Check

```python
# 1) check entire dataframe for nulls
```

```python
print(df.isnull())
```

```python
print(df.isna())
```

```python
# 2) column-wise null count
```

```python
print(df.isnull().sum())
```

```python
# 3) row-wise null count
```

```python
print(df.isnull().sum(axis=1))
```

```python
# 4) Total Null Count (entire dataframe)
```

```python
print(df.isnull().sum().sum())
```

```python
# 5) null percentage by column
```

```python
print((df.isnull().sum() / len(df)) * 100)
```

```python
# 6) overall null percentage
```

```python
print((df.isnull().sum() / df.size) * 100)
```

```python
# 7) nulls in a single column 
```

```python
print(df['BloodPressure'].isnull())
```

```python
# 8) count nulls in a single column
```

```python
print(df['BloodPressure'].isnull().sum())
```

```python
# 9)  check selected column nulls
```

```python
print(df[['BloodPressure','SkinThickness']].isnull())
```

```python
# 10) count nulls in a selected columns
```

```python
print(df[['BloodPressure','SkinThickness']].isnull().sum())
```

```python
# 11) check if a column is completely null
```

```python
print(df['SkinThickness'].isnull().all())
```

```python
# 12) rows containing atleast one null
```

```python
print(df[df.isnull().any(axis=1)])
```

```python
# 13) entire columns with null
```

```python
print(df.isnull().all())
```

```python
# 14) row having all nulls
```

```python
print(df.isnull().all(axis=1))
```

```python
# 15) rows without nulls
```

```python
print(df.notnull().all())
```

```python
# 16) not null check any
```

```python
print(df.notnull().any())
```
